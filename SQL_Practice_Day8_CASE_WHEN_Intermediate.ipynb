## Q1

**Task:** Create employee grades: A (>=90000), B (70000-89999), C (<70000).

```python
import pandas as pd
df=pd.DataFrame({'emp_id':[1,2,3,4],'salary':[95000,82000,65000,72000]})
df
```

```python
import pandasql
from pandasql import sqldf

sqldf("""
select emp_id ,salary,
        case
            when salary >= 90000 then "A"
            when salary >= 70000 and salary  <=89999 then "B"
            else "C"
        end as "grade"
from df



""")
```

## Q2

**Task:** Count male and female employees using CASE + SUM().

```python
import pandas as pd
df=pd.DataFrame({'emp_id':[1,2,3,4,5],'gender':['M','F','M','F','M']})
df
```

```python
sqldf("""

SELECT
    SUM(CASE WHEN gender = 'M' THEN 1 ELSE 0 END) AS male_count,
    SUM(CASE WHEN gender = 'F' THEN 1 ELSE 0 END) AS female_count
FROM df;
""")
```

## Q3

**Task:** Calculate total sales for Online and Offline channels separately.

```python
import pandas as pd
df=pd.DataFrame({'channel':['Online','Offline','Online','Offline'],'amount':[1000,800,1200,900]})
df
```

```python
sqldf("""
SELECT 
SUM(CASE WHEN CHANNEL = 'Online' THEN AMOUNT ELSE  0  END ) AS ONLINE_CAT,
SUM(CASE WHEN CHANNEL = 'Offline' THEN AMOUNT ELSE 0 END ) AS OFFLINE
FROM df


""")
```

## Q4

**Task:** Find average salary of HR and Non-HR employees using CASE.

```python
import pandas as pd
df=pd.DataFrame({'department':['HR','HR','IT','Sales'],'salary':[50000,60000,70000,80000]})
df
```

```python
sqldf("""
SELECT 
AVG(CASE WHEN DEPARTMENT = "HR" THEN SALARY ELSE 0 END) AS AVG_HR_SALARY,
AVG(CASE WHEN DEPARTMENT != "HR" THEN SALARY ELSE 0 END) AS NON_HR_SALARY
FROM df


""")
```
