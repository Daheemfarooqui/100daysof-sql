## Question 1

Extract the year from orders.order_date. Return order_id and year.

```python
import pandasql
from pandasql import sqldf

sqldf("""
select order_id , strftime('%Y', order_date) as order_year
from orders

""")
```

## Question 10

## Question 11

Calculate total order amount for each quarter.

```python
sqldf("""
   SELECT
    strftime('%Y', order_date) AS year,
    ((CAST(strftime('%m', order_date) AS INTEGER) - 1) / 3) + 1 AS quarter,
    SUM(amount) AS total_order_amount
FROM orders
GROUP BY year, quarter
ORDER BY year, quarter;
""")
```

## Question 20

Classify orders as Weekend or Weekday using the extracted day-of-week. Return order_id, order_date and day_type.

```python
sqldf("""
with my_cte as (
    select *,
            strftime('%w',order_date) as day_type
    from orders

)
select 
order_id,order_date , day_type,
case 
    when day_type in [0,5,6] then 'weekend'
    else 'weekday'
end as classify
        
from my_cte
""")
```
