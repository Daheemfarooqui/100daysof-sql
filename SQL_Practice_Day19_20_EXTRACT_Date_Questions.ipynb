## Question 1

Extract the year from order_date and count completed orders for each year.

```python
import pandasql
from pandasql import sqldf
sqldf("""
select strftime('%Y',order_date) as order_year , count(*) as completed
from orders
group by customer_id
having status == 'Completed'




""")
```

## Question 2

Extract the month number from order_date and calculate completed revenue for each month.

```python
sqldf("""
select strftime('%m',order_date) as order_month, sum(amount) as total_revenue
from orders
group by customer_id
having status == 'Completed'

""")
```

## Question 3

Extract the quarter from order_date and calculate the number of orders in each quarter.

```python
sqldf("""
    select 
    ((cast(strftime('%m',order_date)as integer) -1)/3)+1 as quarter,
    count(*) as total_orders
    from orders
    group by quarter


""")
```

## Question 4

Find the highest-value completed order in each quarter. Return quarter, order_id and amount.

```python
sqldf("""
    select
    ((cast(strftime('%m',order_date) as integer) -1)/3)+1 as quarter,
    order_id,
    max(amount) as higest_value_amout
    from orders
    group by quarter
    order by quarter


""")
```

## Question 5

## Question 9

Find the month with the highest completed revenue.

```python
sqldf("""
    select
    strftime('%m',order_date) as order_month,
    sum(amount) as total_amount
    from orders
    group by order_month
    having sum(amount) = (select max(sum_amount)
                            from (select strftime('%m',order_date) as order_month,
                                        sum(amount) as sum_amount
                                    from orders
                                    group by order_month))
                                    

""")
```

## Question 10

Find the day of the month on which the most orders were placed.

```python
sqldf("""
select day_month ,total_orders,rn
from (select 
        strftime('%d',order_date) as day_month,
        count(*) as total_orders,
        dense_rank()
        over(order by count(*) desc)as rn
        from orders
        group by day_month)t
where rn = 1


""")
```

## Question 12

Calculate completed revenue separately for weekdays and weekends.

```python
sqldf("""
select 
case
    when cast(strftime('%w',order_date)as integer) in (0,6) then 'Weekend'
    else 'weekday'
end as day_type,
sum(amount) as total_revenue
from orders
where status = 'Completed'
group by day_type


""")
```

## Question 14

Find the hour of the day with the highest total payment amount.

```python
sqldf("""
select strftime('%H',payment_time) as paymnet_hour,
        sum(amount) as total_amout
from payments
group by paymnet_hour
having sum(amount) = (select max(total_payment)
                        from (
                                select strftime('%H',payment_time) as paymnet_hour,
                                sum(amount) as total_payment
                                from payments
                                group by paymnet_hour)
)
""")
```

## Question 19

Classify orders as Small (<2000), Medium (2000–3500), or Large (>3500), and also extract the order month.

```python
sqldf("""
select 
strftime('%m',order_date) as order_month,

case 
    when sum(amount) < 2000 then 'Small'
    when sum(amount) >=2000 and sum(amount) <=3500 then 'Medium'
    else 'Large'
end as Classify
from orders
group by order_month


""")
```

## Question 20

Find the first and last order month in the dataset using extracted year and month.

```python
sqldf("""
SELECT
    MIN(strftime('%Y-%m', order_date)) AS first_order_month,
    MAX(strftime('%Y-%m', order_date)) AS last_order_month
FROM orders
""")
```
