## Q1

**Task:** Classify employees as High Salary (>=70000) else Low Salary.

```python
import pandas as pd
df=pd.DataFrame({'emp_id':[1,2,3,4],'name':['A','B','C','D'],'salary':[50000,75000,65000,90000]})
df
```

```python
import pandasql
from pandasql import sqldf

sqldf("""
select emp_id,
        name,
        salary,
        
case 
     when salary >= 70000 then "High Salary"
     else "Low Salary"
end as salary_buckat
from df


""")
```

## Q2

**Task:** Classify loan amount as Small, Medium, Large.

```python
import pandas as pd
df=pd.DataFrame({'loan_id':[1,2,3,4],'amount':[25000,75000,150000,50000]})
df
```

```python
sqldf("""
    select 
    loan_id,
    amount,
        case
            when amount >= 100000 then "Large"
            when amount >= 50000   then "medium"
            when amount >= 10000   then "low"
        end as "classify"
    from df



""")
```

## Q3

**Task:** Return Pass if marks>=35 else Fail.

```python
import pandas as pd
df=pd.DataFrame({'student':['A','B','C','D'],'marks':[30,35,70,22]})
df
```

```python
sqldf(""" 
    select
        student,
        marks,
            case
                when marks >= 35 then "Pass"
                else "Fail"
            end as "result"
    from df



""")
```

## Q4

**Task:** Assign bonus based on department.

```python
import pandas as pd
df=pd.DataFrame({'name':['A','B','C'],'department':['Sales','HR','IT']})
df
```

```python
sqldf(""" 
select
    name,
    department,
        case 
            when department = 'Sales' then "20%"
            when department = 'HR'    then "15%"
            when department = 'IT'    then "10%"
        end as "Bonus"
from df

""")
```

## Q5

**Task:** Create amount category: High (>100000) else Normal.

```python
import pandas as pd
df=pd.DataFrame({'customer_id':[101,102,103],'amount':[50000,120000,80000]})
df
```

## Q6

**Task:** Count High and Low salary employees using CASE.

```python
import pandas as pd
df=pd.DataFrame({'salary':[50000,75000,65000,90000,45000]})
df
```

```python
sqldf("""
select salary,
    CASE
        WHEN SALARY > 50000 THEN "HIGH"
        ELSE "LOW"
    END AS "BUCKAT"
FROM df



""")
```

## Q7

**Task:** Customer category: Gold, Silver, Bronze.

```python
import pandas as pd
df=pd.DataFrame({'customer_id':[101,102,103,104],'spending':[1200,900,300,600]})
df
```

```python
sqldf("""
SELECT CUSTOMER_ID , SPENDING,
    CASE 
        WHEN SPENDING >=1000 THEN "GOLD"
        WHEN SPENDING >= 500 THEN "SLIVER"
        ELSE "BRONZE"
    END AS "CATEGORY"
FROM df


""")
```

## Q8

**Task:** Replace NULL department with Unknown using CASE.

```python
import pandas as pd
import numpy as np
df=pd.DataFrame({'name':['A','B','C'],'department':['HR',np.nan,'IT']})
df
```

```python
sqldf("""
SELECT NAME, DEPARTMENT,
    CASE
        WHEN DEPARTMENT IS NULL THEN "UNKNOWN"
        ELSE "DEPARTMENT"
    END AS "DEPARTMENT"
FROM df


""")
```
