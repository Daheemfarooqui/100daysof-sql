## Question 1

Write a SQL query to find the second highest salary from the Employees table.

```python
sqldf("""
select employee_name , salary
from (select employee_name , salary,
        dense_rank()
        over(order by salary desc) as rn
        from employee_salary) t
where rn = 2



""")
```

## Question 2

Write a SQL query to find duplicate records in `duplicate_records` using the same customer_id, email, and amount.

```python
sqldf("""
select customer_id , count(*) as duplicate_record
from duplicate_records
group by customer_id , email, amount
having count(*) > 1 


""")
```

## Question 5

Write a SQL query to find the Nth highest salary. Make the logic usable for an N such as 3 or 5.

```python
sqldf("""

select * 
from (select * ,
        dense_rank()
        over(order by salary desc ) as rn
        from employee_salary) t
where rn <= 3
""")
```

## Question 18

Write a SQL query to find customers who made at least one purchase in every month represented in the orders table.

```python
sqldf("""
SELECT customer_id
FROM orders
GROUP BY customer_id
HAVING COUNT(DISTINCT strftime('%Y-%m', order_date)) =
(
    SELECT COUNT(DISTINCT strftime('%Y-%m', order_date))
    FROM orders
)
""")
```

## Question 20

## Question 21

Write a SQL query to find employees earning more than the average salary of their own department.

```python
sqldf("""
select *
from employee_salary a
where salary > (select avg(salary) 
                from employee_salary b
                where a.department_id = b.department_id)


""")
```
