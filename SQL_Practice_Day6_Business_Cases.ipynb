## Q1

**Business Case:** Find the top 5 customers by total spending.

```python
import pandas as pd
df=pd.DataFrame({'customer_id':[101,101,102,103,104,105,106],'amount':[100,300,500,700,600,400,800]})
df
```

```python
import pandasql
from pandasql import sqldf
sqldf("""
select customer_id , sum(amount) as total_spending
from df
group by customer_id
order by total_spending desc
limit 5



""")
```

## Q5

**Business Case:** Find customers whose spending is above the average customer spending.

```python
import pandas as pd
df=pd.DataFrame({'customer_id':[101,101,102,103,104],'amount':[100,400,500,700,300]})
df
```

```python
sqldf("""
SELECT customer_id,
       SUM(amount) AS total_spending
FROM df
GROUP BY customer_id
HAVING SUM(amount) > (
    SELECT AVG(sum_amount)
    FROM (
        SELECT customer_id,
               SUM(amount) AS sum_amount
        FROM df
        GROUP BY customer_id
    ) t
)


""")
```

## Q6

**Business Case:** Find the second highest salary in each department.

```python
import pandas as pd
df=pd.DataFrame({'name':['A','B','C','D','E','F'],'department':['HR','HR','HR','IT','IT','IT'],'salary':[50000,60000,70000,40000,45000,80000]})
df
```

```python
sqldf("""
select name , department , salary
from (
    select name , department , salary,
    dense_rank()
    over(partition by department order by salary desc) as rn
    from df ) t
where rn = 2

""")
```

## Q10

## Q11

**Business Case:** Find repeat customers (ordered on different dates).

```python
import pandas as pd
df=pd.DataFrame({'customer_id':[101,101,102,103,103],'order_date':['2025-01-01','2025-02-01','2025-01-05','2025-01-10','2025-03-10']})
df
```

```python
sqldf("""
    select customer_id , order_date, rn
    from (select customer_id , order_date,
         row_number()
         over(partition by order_date) as rn
           from df)



""")
```

## Q12

**Business Case:** Find category generating the highest revenue.

```python
import pandas as pd
df=pd.DataFrame({'category':['A','A','B','B','C'],'amount':[100,300,500,400,900]})
df
```

```python
sqldf("""
select category  ,sum(amount) as higest_rev
from df
group by category
having sum(amount) = (select max(higest_rev)
                        from (select sum(amount) as higest_rev
                                from df
                                group by category))
""")
```

## Q13

## Q15

**Business Case:** Find customers who spent less than the average customer spending using a CTE.

```python
import pandas as pd
df=pd.DataFrame({'customer_id':[101,101,102,103,104],'amount':[100,400,500,700,300]})
df
```

```python
sqldf("""
with my_cte as (
    select customer_id ,sum(amount) as total_spend
    from df
    group by customer_id
)

select * from my_cte
where total_spend < (select avg(total_spend)
                        from my_cte)

""")
```
